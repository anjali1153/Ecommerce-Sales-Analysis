```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
plt.rcParams['figure.figsize'] = (10,5)
plt.rcParams['figure.dpi'] = 150
sns.set_theme(style = 'darkgrid',palette = 'twilight')
```

```python
df = pd.read_csv('amazon.csv')
df
```

```python
df.shape
```

```python
df.info()
```

```python
# discounted_price,actual_price,discount_percentage,rating,rating_count this numerical column shows as object/text column changing it into int

df['discounted_price'] = df['discounted_price'].str.replace('₹','').str.replace(',','').astype(float)

df['actual_price'] = df['actual_price'].str.replace('₹','').str.replace(',','').astype(float)

df['rating'] = pd.to_numeric(df['rating'],errors = 'coerce')    # This will replace '|' or any non-number with NaN.

df['rating_count'] = df['rating_count'].str.replace(',','')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce').fillna(0).astype(int)
```

```python
df.info()
```

```python
df['discount_percent_cal'] = (((df['actual_price']-df['discounted_price'])/df['actual_price'])*100).round(2)
```

```python
# Average Product Rating by Category

average_rating = df.groupby('category')['rating'].mean().sort_values(ascending = False)
average_rating
```

```python
#verage Discount by Category
avg_discount = df.groupby('category')['discount_percent_cal'].mean().sort_values(ascending=False)
avg_discount
```

```python
# Visualizations
# Top Categories (Bar Chart)
category_count = df['category'].value_counts().head(10)
category_count.plot(kind='bar')
plt.title("Top 10 Product Categories by Count")
plt.xlabel("Category")
plt.ylabel("Number of Products")
plt.savefig("Top 10 Product Categories by Count.png",bbox_inches='tight')
plt.show()
```

```python
average_rating.head().plot(kind='pie',autopct='%0.02f')
plt.title("Top 5 Average rating  by category")
plt.savefig("Top 5 Average rating  by category.png", bbox_inches='tight')
```

```python
sns.kdeplot(df['rating'].dropna(), shade=True)
plt.title("Rating Distribution (KDEplot)")
plt.xlabel("Rating")
plt.ylabel("Density")
plt.savefig("rating_kde_distribution.png", bbox_inches='tight')
plt.show()
```
